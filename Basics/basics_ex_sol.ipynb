### Exercise 2

```python
def factorial(n):
    if n < 0:
        return 0
    result = 1
    while n > 0:
        result *= n
        n -= 1
    return result

# BONUS
import math

def exp_taylor(x, n_terms):
    total = 0.
    for k in range(n_terms):
        total += x**k/factorial(k)
    return total
```

```python
print(factorial(5))
approx = exp_taylor(1., 12)
print(approx)
# Compare with the exact value: use isclose, NOT ==
print(approx == math.exp(1.), math.isclose(approx, math.exp(1.)))
```

### Exercise 3

```python
def count_defects(log):
    counts = {}
    for defect in log:
        if defect in counts:
            counts[defect] += 1
        else:
            counts[defect] = 1
    return counts

# Alternative, using the get() method with a default value:
def count_defects_get(log):
    counts = {}
    for defect in log:
        counts[defect] = counts.get(defect, 0) + 1
    return counts
```

```python
print(count_defects(["crack", "corrosion", "crack", "wear", "corrosion", "crack"]))
print(count_defects_get(["crack", "corrosion", "crack", "wear", "corrosion", "crack"]))
```

### Exercise 4

```python
def count_letters_digits(code):
    letters = 0
    digits = 0
    for char in code:
        if char.isalpha():
            letters += 1
        elif char.isdigit():
            digits += 1
    return letters, digits
```

```python
letters, digits = count_letters_digits("AL-6061-T6")
print(letters, digits)
```

### Exercise 5

```python
# A set discards the duplicates, sorted() puts the result in ascending order.
# No loop is needed, so this is already the BONUS version.
def unique_loads(loads):
    return sorted(set(loads))
```

```python
print(unique_loads([1, 4, 2, 5, 3, 4, 3, 5, 200, 10]))
```

### Exercise 6

```python
def is_symmetric(cycle):
    # Remove the spaces, then split on the commas
    steps = cycle.replace(" ", "").split(",")
    return steps == steps[::-1]

# BONUS
def is_symmetric_many(*cycles):
    return [is_symmetric(cycle) for cycle in cycles]
```

```python
print(is_symmetric("0, 5, 10, 5, 0"))
print(is_symmetric("0, 5, 10, 8, 0"))
print(is_symmetric_many("0, 5, 10, 5, 0", "0, 5, 10, 8, 0"))
```

### Exercise 7

```python
import math

def cross_section_areas(diameters):
    return [math.pi*d**2/4 for d in diameters]

# BONUS: the same, with map and a lambda
def cross_section_areas_map(diameters):
    return list(map(lambda d: math.pi*d**2/4, diameters))
```

```python
areas = cross_section_areas([10., 12., 16.])
print(areas)
print([round(a, 2) for a in areas])
print(cross_section_areas_map([10., 12., 16.]))
```

### Exercise 8

```python
# The bug: "data" is only an ALIAS of the caller's list, so assigning to data[i]
# modifies the caller's list in place. The fix is to build a new list instead
# (a copy, data[:], would work as well).
def normalize(data):
    peak = max(data)
    return [value/peak for value in data]
```

```python
readings = [2., 4., 8., 1.]
print("returned:", normalize(readings))
print("readings:", readings)   # unchanged, as it should be
```

### Exercise 9

```python
import math

def sqrt_newton(a, tol=1e-12, max_iter=100):
    x = a
    n_iter = 0
    while n_iter < max_iter:
        x_new = 0.5*(x + a/x)
        n_iter += 1
        if math.isclose(x_new, x, rel_tol=tol):
            x = x_new
            break
        x = x_new
    return x, n_iter
```

```python
root, n_iter = sqrt_newton(2.)
print(root, n_iter)
# The result is NOT bit-for-bit identical to math.sqrt, but it is close enough
print(root == math.sqrt(2.), math.isclose(root, math.sqrt(2.)))
```

### Exercise 11

```python
def velocities(times, positions):
    result = []
    # zip pairs each sample with the following one
    for (t1, t2), (x1, x2) in zip(zip(times, times[1:]), zip(positions, positions[1:])):
        t_mid = (t1 + t2)/2
        v = (x2 - x1)/(t2 - t1)
        result.append((t_mid, v))
    return result
```

```python
result = velocities([0., 0.1, 0.2, 0.3], [0., 1.2, 4.9, 11.])
print(result)
# Rounded, to hide the usual floating-point noise
print([(round(t, 2), round(v, 2)) for t, v in result])
```

### Exercise 12

```python
database = {"steel":     {"density": 7850, "E": 210e9},
            "aluminium": {"density": 2700, "E": 69e9}}

def get_property(database, material, prop, default="unknown"):
    # The first get() returns an empty dictionary if the material is missing,
    # so the second get() can always be called safely.
    return database.get(material, {}).get(prop, default)
```

```python
print(get_property(database, "steel", "density"))
print(get_property(database, "copper", "density"))
print(get_property(database, "steel", "poisson", 0.3))

# Dictionary comprehension: convert every density to g/cm^3
densities = {name: props["density"]/1000 for name, props in database.items()}
print(densities)
```

### Exercise 13

```python
def three_densest(materials):
    # key tells sorted() what to sort by: here, the second item of each tuple
    return sorted(materials, key=lambda material: material[1], reverse=True)[:3]
```

```python
print(three_densest([("steel", 7850), ("aluminium", 2700), ("titanium", 4500),
                     ("copper", 8960), ("magnesium", 1740)]))
```

### Exercise 14

```python
def summarize(data, k=2.):
    n = len(data)
    mean = sum(data)/n
    std = (sum([(value - mean)**2 for value in data])/n)**0.5
    kept = [value for value in data if abs(value - mean) <= k*std]
    return mean, std, kept, n - len(kept)
```

```python
mean, std, kept, discarded = summarize([2.1, 2.3, 2.2, 9.8, 2.0, 2.4, 2.1])
print(f"mean = {mean:.3f}, std = {std:.3f}")
print("kept:", kept)
print("discarded:", discarded)
```

### Exercise 15

```python
def report(readings, **options):
    # Read the options out of the kwargs dictionary, with defaults
    threshold = options.get("threshold", 5.)
    invalid = options.get("invalid", -999.)
    for index, value in enumerate(readings):
        if value == invalid:
            continue   # skip the failed readings
        if value > threshold:
            print(f"reading {index}: {value:.1f} exceeds {threshold:.1f}")
```

```python
readings = [2.1, -999., 5.6, 2.0, -999., 7.9]
report(readings)
print("---")
report(readings, threshold=7.)
```
