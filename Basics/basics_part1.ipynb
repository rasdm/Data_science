#### A note on floating-point equality

Floating-point numbers are stored in binary, with a *finite* number of bits (64 in Python, of which 53 for the significant digits), and most decimal fractions have no exact binary representation. Just as $1/3$ has no exact decimal representation ($0.3333\ldots$), $0.1$ in binary is the periodic fraction $0.0001100110011\ldots_2$, which has to be truncated to fit in 53 bits. What is actually stored is therefore the closest representable number to $0.1$, not $0.1$ itself.

Each operation on such values then rounds its result to the nearest representable number, and these tiny errors (of relative size $\sim 10^{-16}$) accumulate. This is why results sometimes come out as `0.30000000000000004` instead of `0.3` (see the cell below), or as `3.3000000000000003` instead of `3.3` when computing `1.1*3`: the extra digits are the accumulated representation error, not a mistake in the computation. As a consequence, the same real number computed in two different ways is not always bit-for-bit identical, and `==` can behave unexpectedly on floats.

```python
# Looks like it should be 0.3...
0.1 + 0.2
```

```python
# ...but it isn't, exactly!
0.1 + 0.2 == 0.3
```

For this reason, never use `==` to compare floats for (near-)equality. Use `math.isclose()` instead (or `numpy.allclose()` for arrays), which checks that two values are within a small tolerance of each other. `math` is a *module* of the standard library: the `import` statement below makes its contents available under the name `math` (modules and imports are covered in Part 2).

```python
import math
math.isclose(0.1 + 0.2, 0.3)
```

**Slicing**: accessing a portion of the elements in a list. It follows the general format: lst[initial_index:final_index:step].
```python
# The list without the first element (omitting final index and step)
a[1:]
```

```python
# This slice is equivalent to the previous one, without omitted indices
a[1:5:1]
```

```python
# The list without the last element (notice that the final index is excluded from the slice)
a[:-1]
```

```python
# Get even-indexed elements (omitting initial and final indices)
a[::2]
```

```python
# Reverse a list
a[::-1]
```

### Mutability and aliasing

When you assign a *mutable* object (like a list) to a new variable, you are **not** making a copy: both variable names now point to the very same object in memory. This is called *aliasing*. Mutating the object through either name is visible through the other name too.

```python
# nums and same refer to the SAME list object (aliasing, not a copy)
nums = [1, 2, 3]
same = nums
same.append(4)
print("nums:", nums)
print("same:", same)
```

This explains what happened earlier with `aa = [a, a[1:]]`. `aa[0]` is an *alias* of `a` (the same list object), while `aa[1]` was built from `a[1:]`, a slice, which always creates a **new** (copied) list. Since then, we called `a.append("Odense")` and `a.pop(0)`, which mutated `a` in place. Let's look at `aa` again:

```python
aa
```

Notice that `aa[0]` reflects the mutations we made to `a`, while `aa[1]` is unchanged, since it was a separate list all along. To make an explicit, independent copy of a list, use slicing (`a[:]`), `list(a)`, or `a.copy()`:

```python
original = [1, 2, 3]
copy1 = original[:]   # equivalent to list(original) or original.copy()
copy1.append(4)
print("original:", original)
print("copy1:", copy1)
```

Beware also of *nested* structures: a shallow copy (`a[:]`, `list(a)`, `.copy()`) only copies the outer container: the inner lists are still shared. Use `copy.deepcopy()` to copy everything recursively.

```python
import copy

nested = [[1, 2], [3, 4]]

# A shallow copy only copies the OUTER list: the inner lists are still shared
shallow = nested[:]
shallow[0].append(99)
print("after mutating 'shallow':")
print("  nested: ", nested)   # changed too, even though we only touched 'shallow'!
print("  shallow:", shallow)

# A deep copy copies everything recursively, so the inner lists are independent
deep = copy.deepcopy(nested)
deep[0].append(-1)
print("after mutating 'deep':")
print("  nested:", nested)    # this time nested does NOT see the -1
print("  deep:  ", deep)
```

The same aliasing behaviour also applies to the arguments of a *function*: we will come back to it in Part 2, once functions have been introduced.

## Strings

Strings are *immutable* sequences of characters: like lists and tuples, they support indexing, slicing, `len()`, and the `in` operator, but they also come with their own set of useful methods.

```python
s = "  Aarhus University  "
# Indexing and slicing work as for lists/tuples
print(s[2])
print(s[2:9])
```

#### Membership: the `in` operator

`in` checks whether a value is contained in a string (substring check), list, tuple, set, or dictionary (checks the *keys*). It returns a boolean.

```python
print("Aarhus" in "Aarhus University")     # substring check
print(2 in [1, 2, 3])                       # list membership
print("key_1" in {"key_1": 1, "key_2": 2})  # dict membership checks the keys
```

### Format specifiers

After the value, a colon introduces a *format specifier* that controls how the value is displayed. This is essential when reporting numerical results: it lets you fix the number of decimals instead of printing all the digits stored in memory.

Some common specifiers:

| Specifier | Meaning |
|---|---|
| `.3f` | fixed-point notation with 3 decimals |
| `.2e` | scientific (exponential) notation with 2 decimals |
| `8.2f` | fixed-point, 2 decimals, padded to a total width of 8 characters |
| `.1%` | percentage with 1 decimal |
| `,` | thousands separator |

```python
x = 2/3

print(f"Default:      {x}")
print(f"3 decimals:   {x:.3f}")
print(f"Scientific:   {x:.2e}")
print(f"As a percent: {x:.1%}")
```

Width specifiers are useful to align values in columns, e.g. when printing a table of results. The number after the colon sets the minimum width of the field; `>` right-aligns the value inside that width (`<` left-aligns it, which is the default for strings).

```python
# Padding each value to a fixed width makes the columns line up
print(f"{'material':12}{'stress [Pa]':>14}{'strain':>10}")
print(f"{'steel':12}{355e6:14.2e}{0.00218:10.4f}")
print(f"{'aluminium':12}{95e6:14.2e}{0.00417:10.4f}")
```

### Older formatting styles

f-strings were introduced in Python 3.6. You will still meet two older styles in existing code, so it is useful to recognise them, but prefer f-strings in your own code. The same format specifiers work with all three styles.

```python
# The .format() method: values are passed as arguments and fill the {} placeholders in order
print("I'm going to add {} here.".format('something'))
print("{} divided by {} is {:.2f}".format(2, 3, 2/3))
```

```python
# The oldest style, inherited from the C language: the % operator with %s, %d, %f placeholders
print("I'm going to add %s here." % 'something')
print("2 divided by 3 is %.2f" % (2/3))
```
