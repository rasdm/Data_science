   $$\sigma_{eq} = \sqrt{\sigma_1^2 - \sigma_1\sigma_2 + \sigma_2^2}.$$

   ```python
   s1 = np.array([120., -80., 200.,  50.])   # MPa
   s2 = np.array([ 40., -150., -60.,  50.])  # MPa
   ```

6. The free vibration of a damped single-degree-of-freedom system has amplitude

   with $A_0 = 5$ mm, $\zeta = 0.05$ and $\omega_n = 2\pi\cdot 20$ rad/s.

   1. Build an array of $6$ instants between $0$ and $0.5$ s and compute the corresponding amplitudes with `np.exp`.
   2. Taking the logarithm of the amplitudes (`np.log`), verify that they decay linearly in time: the successive differences of $\log A$ must all be equal (use `np.allclose`).
   3. Recover the damping ratio $\zeta$ from the slope of $\log A$ and check that it matches the value used to generate the data.

````python
A0 = 5.               # mm
zeta = 0.05
omega_n = 2*np.pi*20  # rad/s

# Your code here
t=np.linspace(0,0.5,6)
At=A0*np.exp(-zeta*omega_n*t)

logAt=np.log(At)
print(np.isclose(logAt[0]-logAt[1],logAt[1]-logAt[2]))

slope=(logAt[1]-logAt[0])/(t[1]-t[0])
print(np.isclose(-slope/omega_n,zeta))
````

## Linear algebra

17. Three masses are connected by springs to a wall and to each other. Under the static loads $f$, their displacements $u$ solve the linear system $K u = f$, with

    ```python
    K = 1e3*np.array([[200., -100.,    0.],
                      [-100.,  200., -100.],
                      [   0., -100.,  100.]])   # N/m
    f = np.array([0., 0., 500.])                # N
    ```

    1. Solve the system with `np.linalg.solve` and print the displacements in mm.
    2. Verify the solution: check with `np.allclose` that $Ku = f$ (the product can be written `K @ u` or `np.dot(K, u)`), and print the norm of the residual $\|Ku-f\|$ with `np.linalg.norm`.
    3. Check that $K$ is symmetric, comparing it with its transpose.
    4. Compute $K^{-1}$ and verify that $K K^{-1}$ is the identity matrix (`np.eye`). Note that solving the system directly is preferable to inverting the matrix.

    BONUS. If the three masses are unitary, the natural angular frequencies of the system are $\omega_j = \sqrt{\lambda_j}$, where $\lambda_j$ are the eigenvalues of $K$. Compute them with `np.linalg.eigh` and print the corresponding frequencies $f_j = \omega_j/(2\pi)$ in Hz.

````python
K = 1e3*np.array([[200., -100.,    0.],
                  [-100.,  200., -100.],
                  [   0., -100.,  100.]])   # N/m
f = np.array([0., 0., 500.])                # N

# Your code here
u=np.linalg.solve(K,f)
print(u)
print(np.allclose(K @ u,f))
print(np.linalg.norm(K @ u - f))

print(np.allclose(K,K.T))

Kinv=np.linalg.inv(K)
print(np.allclose(K@Kinv,np.eye(3,3)))

eigval, eigvec=np.linalg.eigh(K)
freq=np.sqrt(eigval)/(2*np.pi)
print(freq)
````
