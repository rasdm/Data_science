   Write a function that checks whether a cycle is symmetric. Use `replace()` to get rid of the spaces
   ([documentation](https://docs.python.org/3/library/stdtypes.html#str.replace)) and `split(",")` to
   separate the levels.

   Sample input: `"0, 5, 10, 5, 0"`. Expected output: `True`

   Sample input: `"0, 5, 10, 8, 0"`. Expected output: `False`

   BONUS: write a function that checks an *arbitrary* number of cycles (variable number of arguments).

   Sample input: two arguments = `"0, 5, 10, 5, 0"`, `"0, 5, 10, 8, 0"`. Expected output: `[True, False]`

````python
def is_symmetric(cycle):
    cycle=cycle.replace(" ", "").split(",")
    return cycle==cycle[::-1]

def symmetric_args(*cycles):
    return [is_symmetric(cycle) for cycle in cycles]
    
sample="0, 5, 10, 5, 0"
sample2="0, 5, 10, 8, 0"
print(is_symmetric(sample))
print(symmetric_args(sample, sample2))

````

9. The square root of $a$ can be computed with the iteration $x_{n+1}=\frac{1}{2}\left(x_n+a/x_n\right)$,
   starting from $x_0=a$. Write a function that runs this iteration in a `while` loop and stops (`break`)
   as soon as two successive iterates are close, or after `max_iter` iterations, whichever comes first.
   Give `tol` and `max_iter` default values. Return both the root and the number of iterations performed.

   Use `math.isclose` to test convergence: with `==` the loop would essentially never stop (see Part 1).

   Sample input: `a = 2.0`. Expected output: `(1.414213562373095, 6)`

   Note that this differs from `math.sqrt(2.0)` in the last digit, so `result == math.sqrt(2.0)` is `False`
   while `math.isclose(result, math.sqrt(2.0))` is `True`.

````python
import math

def sqrt_newton(a, tol=1e-12, max_iter=100):
    x=a-2*tol
    x_plus=a
    i=0
    while abs(x - x_plus) > tol and i < max_iter:
        x=x_plus
        i+=1
        x_plus=1/2*(x+a/x)
    return x_plus, i

print(sqrt_newton(2))
````

11. A sensor recorded the position of a slider at successive times. Write a function that estimates the
    average velocity over each interval, $v=(x_{i+1}-x_i)/(t_{i+1}-t_i)$, and returns a list of tuples
    `(t_mid, v)` where `t_mid` is the midpoint of the interval. TIP: `zip(times, times[1:])` pairs each
    sample with the following one.

    Sample input: `times = [0.0, 0.1, 0.2, 0.3]`, `positions = [0.0, 1.2, 4.9, 11.0]`.
    Expected output (rounded to 2 decimals): `[(0.05, 12.0), (0.15, 37.0), (0.25, 61.0)]`

````python
def velocities(times, positions):
    result=[]
    for (t1, t2), (x1, x2) in zip(zip(times, times[1:]),zip(positions, positions[1:])):
        t_mid=(t1+t2)/2
        v=(x2-x1)/(t2-t1)
        result.append((t_mid, v))
    return result

times = [0.0, 0.1, 0.2, 0.3]
positions = [0.0, 1.2, 4.9, 11.0]
res=velocities(times, positions)

print([(round(t,2), round(v,2)) for t,v in res])
````
